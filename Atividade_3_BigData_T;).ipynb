```python
import threading
import time

```

```python
finaliza_laco = False # variavel global de controle
def stream_dados(arquivo):
    with open(arquivo, "r") as arq:
        while not finaliza_laco:
            linha = arq.readline().strip()
            if not linha:
                time.sleep(1) # apenas para depuração
                continue
            yield linha
```

```python
def monitora_linhas(arquivo):
    print('inicializando monitoramento')
    for nova_linha in stream_dados(arquivo):
        print(f"processando nova linha:\n\t{nova_linha}")
        pass # vista de tempo real
    print('finalizando monitoramento')
```

```python
finaliza_laco = False
t = threading.Thread(
    target=monitora_linhas,
    args=("fluxo.log",))
t.start()
```

```python
print(t.is_alive()) # True
finaliza_laco = True
time.sleep(1) # período do laço do stream_dados
print(t.is_alive()) # False
```
